```python
%python

import requests

# Step 1: Store the URL
url = "https://public.tableau.com/app/sample-data/mobile_os_usage.csv"

# Step 2: Perform API call
response = requests.get(url)

# Step 3: Check whether the request was successful
response.raise_for_status()

# Step 4: Store the downloaded data in a variable
variable = response

# Step 5: Write data into Databricks Volume
dbutils.fs.put(
    "/Volumes/workspace/default/usage_metrics/mobile_os_usage.csv",
    variable.text,
    overwrite=True
)

print("Data successfully downloaded and saved!")
```

```python
%python

display(dbutils.fs.ls("/Volumes/workspace/default/usage_metrics/"))
df1 = spark.read.option("header", True) \
    .option("inferSchema", True) \
    .csv("/Volumes/workspace/default/usage_metrics/mobile_os_usage.csv")

display(df1)
```

```python
%python

df1 = df1.withColumnRenamed("Mobile Operating System", "Mobile_Operating_System") \
         .withColumnRenamed("Percent of Usage", "Percent_of_Usage")

display(df1)

df1.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.default.mobile_os_usage")

print("Table created successfully!")
```
